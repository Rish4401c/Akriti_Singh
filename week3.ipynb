**Handling Missing values***

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

df= df.dropna(subset=['Date'])
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Calories']=df['Calories'].fillna(df['Calories'].mean())

```

```python
df = df.drop_duplicates()
```

```python
df = df[df['Duration']<= 180]
```

```python
#create new columns

df['Pulse_diff'] = df['Maxpulse'] - df['Pulse']
df['Calories_per_min'] = df['Calories'] / df['Duration']
```

```python
# Line plot: Calories over time

plt.plot(df['Date'], df['Calories'])
plt.xlabel('Date')
plt.ylabel('Calories')
plt.title('Calories Burden Over Time')
plt.show()
```

```python
# Scatter plot: Duration vs Calories
sns.scatterplot(x="Duration", y="Calories", data=df)
plt.title("Calories vs Duration")
plt.show()
```
